```python
from textblob import TextBlob

def correct_grammar(text):
    blob = TextBlob(text)
    corrected_text = str(blob.correct())
    return corrected_text

text = input("Enter your sentence: ")
corrected_text = correct_grammar(text)

print(f"Corrected: {corrected_text}")
```

```python
"""
grammar_corrector.py -- correct grammar in a sentence.

Install:  pip install language-tool-python
The local checker needs Java 17+. If you don't have it, this falls back
to LanguageTool's free hosted API (rate-limited; don't send private text).
"""

import language_tool_python


def make_tool(language="en-US"):
    """Try the local checker; fall back to the hosted API if Java is missing."""
    try:
        return language_tool_python.LanguageTool(language)
    except Exception as error:
        print(f"Local checker unavailable ({error})")
        print("Falling back to the online checker...\n")
        return language_tool_python.LanguageToolPublicAPI(language)


def correct_grammar(tool, text):
    """Return the text with grammar and spelling issues fixed."""
    return tool.correct(text)


tool = make_tool()

while True:
    text = input("Enter your sentence (or press Enter to quit): ").strip()
    if not text:
        break
    print(f"Corrected: {correct_grammar(tool, text)}\n")

tool.close()
```
